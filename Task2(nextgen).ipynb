**IQR method to fine and remove outliers\**

calculating q1 and q3

```python
q1=df['fare'].quantile(0.25)
q3=df['fare'].quantile(0.75)
print(q1,q3)
```

```python
IQR=q3-q1
print(IQR)
```

```python
upper_limit=q3+(1.5*IQR)
lower_limit=q1-(1.5*IQR)
print(upper_limit,lower_limit)
```

```python
df[(df.fare >= lower_limit) & (df.fare <= upper_limit)].shape[0]
```

```python
# Correctly filter the DataFrame to keep only rows within the IQR limits and assign to a new DataFrame
df = df[(df.fare >= lower_limit) & (df.fare <= upper_limit)]
df
```

```python
df.info()
```

```python
sns.displot(data=df,x='fare',kde=True)
print(df['fare'].skew())
print(df['fare'].kurtosis())
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# 1. Correct countplot structure with keyword arguments
sns.countplot(data=df, x='survived', hue='pclass')
plt.show()

# 2. Correct and clean way to get percentage crosstab of survival by pclass
cross_tab = pd.crosstab(df['pclass'], df['survived'])
percentage_crosstab = cross_tab.div(cross_tab.sum(axis=1), axis=0) * 100
print(percentage_crosstab.round(1))
```
